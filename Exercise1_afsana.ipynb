```python
# Exercise 1: Exploratory Data Analysis & Automated Reporting

# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All Libraries Imported Successfully!")
```


# Generate random numerical data

np.random.seed(42)

data = {
    'f1': np.random.randint(10, 100, 10),
    'f2': np.random.randint(10, 100, 10),
    'f3': np.random.randint(10, 100, 10),
    'f4': np.random.randint(10, 100, 10)
}

print("Random Data Generated Successfully!")


```python
import os

print("Current Folder:", os.getcwd())

print("Available Files:")
print(os.listdir())
```

```python
df = pd.DataFrame(data)

print("Original DataFrame:")
display(df)
```

```python

# Scatter Plot between f2 and f3

plt.figure(figsize=(8, 5))

plt.scatter(df['f2'], df['f3'])

plt.xlabel('f2 Values')
plt.ylabel('f3 Values')

plt.title('Relationship between f2 and f3')

plt.grid(True)

plt.show()
```

```python

# Calculate Row-wise Arithmetic Mean

df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

print("Updated DataFrame with Arithmetic Mean:")

display(df)
```

```python

# Export DataFrame to Excel

df.to_excel('Exercise1_Report.xlsx', index=False)

print("Excel Report Generated Successfully!")
```
